```python
nomes = []
telefones = []

def novocadastro():
  nome = input("Digite seu nome: ")
  if nome in nomes:
    print("já existe um cadastro com esse nome.")
    return
  telefone =  input("Digite seu número de telefone: ")
  if telefone in telefones:
    print("Esse número de telefone ja foi cadastrado.")
    return
  nomes.append(nome)
  telefones.append(telefone)

def removercadastro():
  if not nomes:
    print("Nenhuma cadastrado até o momento.")
    return
  qnome = input("Qual o nome da pessoa: ")
  if qnome not in nomes:
    posicao = nomes.index(qnome)
    nomes.pop(posicao)
    telefones.pop(posicao)
    print("cadastro removido com sucesso!!!")
  else:
    print("Esse nome não está na cadastrado.")


def verlista():
  if not nomes:
    print("Nenhuma cadastrado até o momento.")
    return
  for i in range(len(nomes)):
    print(f"Nome: {nomes[i]}")
    print(f"Telefone: {telefones[i]}")
    print()

def editarcadastro():
  editarnome = input("Qual cadastro você deseja editar: ")

  if editarnome not in nomes:
    print("Esse nome não está cadastrado, tente novamente!!!")
    return

    qposicao = nomes.index(editarnome)

    print("Opção de edição")
    print("1 - Nome")
    print("2 - Número")
    print("3 - As duas opções")

    qescolha = int(input("O que você deseja editar: "))

    match qescolha:
      case 1:
        novonome = input("Digite o novo nome: ")

        if novonome in nomes:
          print("Esse nome já está cadastrado, tente novamente!!!")
          return

          nomes[qposicao] = novonome
          print("Nome atualizado com sucesso!")

      case 2:
        novotelefone = input("Digite o novo número: ")

        if novotelefone in telefones:
          print("Esse número já está cadastrado, tente novamente!!!")
          return

          telefones[qposicao] = novotelefone
          print("Número atualizado com sucesso!")

      case 3:
        novonome = input("Digite o novo nome: ")

        if novonome in nomes:
          print("Esse nome já está cadastrado, tente novamente!!!")
          return

          novotelefone = input("Digite o novo número: ")

        if novotelefone in telefones:
          print("Esse número já está cadastrado, tente novamente!!!")
          return

          nomes[qposicao] = novonome
          telefones[qposicao] = novotelefone
          print("Cadastro atualizado com sucesso!")

      case _:
        print("Opção inválida!")

while True:

  opcao = (
"===========Bem-vindo===========",
"_______________________________",
"======Em que posso ajudar======",
"_______________________________",
"1 - novo cadastro",
"2 - remover um cadastro",
"3 - ver lista de cadastros",
"4 - editar cadastro",
"5 - frecha o programa",
"______________________________"
)

  for opcaos in opcao :
    print(opcaos)

  escolha = int(input("qual a sua escolha: "))

  match escolha:

    case 1:
      novocadastro()
    case 2:
      removercadastro()
    case 3:
      verlista()
    case 4:
      editarcadastro()
    case 5:
      print("fechada o programa...")
      break
    case _:
      print("Opção inválida!!!")
```
