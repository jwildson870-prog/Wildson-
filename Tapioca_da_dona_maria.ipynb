```python
print("tapioca da dona maria")

valorp = 0
desconto = 0
valorfinal = 0
troco = 0
formapagamento = ""


def frango():

    while True:

        tpf = 8

        print("tapioca de frango custa R$8,00.")
        quantidade = int(input("quantos você quer: "))

        if quantidade <= 0 or quantidade > 10:
            print("Pedido máxima 10, pedindo mínimo 1.")
            continue
        else:
            global valorp
            valorp += tpf * quantidade
            print("Pedido realizado com sucesso!")
            break


def chocolate():

    tpc = 10

    while True:
        print("tapioca de chocolate custa R$10,00.")
        quantidade = int(input("quantos você quer: "))

        if quantidade <= 0 or quantidade > 10:
            print("Pedido máxima 10, pedindo mínimo 1.")
            continue
        else:
            global valorp
            valorp += tpc * quantidade
            print("Pedido realizado com sucesso!")
            break


def queijo():

    tpq = 6

    while True:
        print("tapioca de queijo custa R$6,00.")
        quantidade = int(input("quantos você quer: "))

        if quantidade <= 0 or quantidade > 10:
            print("Pedido máxima 10, pedindo mínimo 1.")
            continue
        else:
            global valorp
            valorp += tpq * quantidade
            print("Pedido realizado com sucesso!")
            break


def carnedesol():

    tpcs = 15

    while True:
        print("tapioca de carne de sol custa R$15,00.")
        quantidade = int(input("quantos você quer: "))

        if quantidade <= 0 or quantidade > 10:
            print("Pedido máxima 10, pedindo mínimo 1.")
            continue
        else:
            global valorp
            valorp += tpcs * quantidade
            print("Pedido realizado com sucesso!")
            break


def vertotal():
    if valorp == 0:
        print("Você ainda não fez nenhum pedido.")
        return

    print(f"Total de: R$ {valorp:.2f}")


def formapagamento():

    def dinheiro():
        global desconto, valorfinal, troco, formapagamento

        formapagamento = "Dinheiro"

        desconto = valorp * 0.10
        valorfinal = valorp - desconto

        valorentregue = float(input("Valor entregue: R$ "))

        if valorentregue < valorfinal:
            print("Valor insuficiente")
            print("Tente novamente!")
            return

        troco = valorentregue - valorfinal

        if troco == 0:
            print("Não tem troco")
        else:
            print(f"Troco: R$ {troco:.2f}")

        print(f"Valor final: R$ {valorfinal:.2f}")
        print(f"Desconto: R$ {desconto:.2f}")
        print("Pagamento finalizado com sucesso!")
        return

    def pix():
        global desconto, valorfinal, troco, formapagamento

        formapagamento = "Pix"

        desconto = valorp * 0.05
        valorfinal = valorp - desconto
        troco = 0

        print(f"Valor final: R$ {valorfinal:.2f}")
        print(f"Desconto: R$ {desconto:.2f}")
        print("Pagamento finalizado com sucesso!")
        return

    def cartao():
        global valorp, desconto, valorfinal, troco, formapagamento

        formapagamento = "Cartão"
        desconto = 0
        valorfinal = valorp
        troco = 0

        print(f"Total: R$ {valorp:.2f}")
        print("Sem desconto.")
        print("Pagamento finalizado com sucesso!")
        return

    while True:

        opcao = (
            "1 - Dinheiro tem um desconto de 10%",
            "2 - Pix tem um desconto de 5%",
            "3 - Cartão não tem desconto"
        )

        for i in opcao:
            print(i)

        escolha = int(input("Forma de pagamento: "))

        match escolha:
            case 1:
                dinheiro()
                break
            case 2:
                pix()
                break
            case 3:
                cartao()
                break
            case _:
                print("Opção inválida")
                print("Tente novamente!")


def notafiscal():
    print("\n" + "=" * 30)
    print("        NOTA FISCAL")
    print("=" * 30)
    print(f"Valor da compra: R$ {valorp:.2f}")
    print(f"Forma de pagamento: {formapagamento}")
    print(f"Desconto: R$ {desconto:.2f}")
    print(f"Valor final: R$ {valorfinal:.2f}")

    if formapagamento == "Dinheiro":
        print(f"Troco: R$ {troco:.2f}")

    print("=" * 30)
    print("Obrigado pela preferência!")
    print("=" * 30)


while True:

    opcao = (
        "1 - Frango",
        "2 - Chocolate",
        "3 - Queijo",
        "4 - Carne de sol",
        "5 - Ver valor até o momento",
        "6 - Fazer pagamento e finalizar pedido"
    )

    for i in opcao:
        print(i)

    escolha = int(input("Qual recheio você deseja: "))

    match escolha:
        case 1:
            frango()
        case 2:
            chocolate()
        case 3:
            queijo()
        case 4:
            carnedesol()
        case 5:
            vertotal()
        case 6:
            formapagamento()
            notafiscal()
            print("Pedido finalizado.")
            break
        case _:
            print("Opção inválida!")
```
